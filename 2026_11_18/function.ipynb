# 參數(parameter)，引數(argument)->真正的值

```python
def greet(name):
    print(f"哈囉,{name}!")

greet("小明")
```

### 引數位置{Position Argument}

```python
def describe_pet(animal, name):
    """顯示寵物資訊"""
    print(f"\n我有一隻{animal}.")
    print(f"我的{animal}叫做{name.title()}.")
    print(f"我養了一隻{animal}，它的名字叫{name}。")

#呼叫
describe_pet("小花", "貓")
describe_pet("貓", "小花")
```

### 引數名稱(Keyword Argument)

```python
def describe_pet(animal, name):
    print(f"我有一隻{animal}，它的名字叫{name}。")

describe_pet(animal="貓", name="小花")
describe_pet(name="小花", animal="貓")
```

## 重要規則:混用時「位置引數」一定要在前面

```python
# 正確:先傳引數位置，再傳引數名稱
describe_pet("狗", name="旺財")

# 錯誤: 位置引數不能跟在引數名稱後面
# describe_pet(animal="狗", "旺財")
```

## 搭配預設值 (Default Arguments)   
定義函式時，如果有幫參數設定預設值，呼叫時就能只改自己想調整的項目：   

```python
def order_drink(drink, ice="正常冰", sugar="半糖"):
    print(f"點餐: {drink}({ice}/{sugar})")

order_drink("紅茶")
order_drink("紅茶", "去冰")
order_drink("紅茶", sugar="半糖")
#錯誤order_drink("紅茶", "半糖")        
order_drink("綠茶", sugar="微糖", ice="少冰") #有定義就不用按照順序
```

## 特殊符號 / 與 * — 限制引數的呼叫方式
### /(僅限位置引數 Positional-Only)   
寫在 '/' 左邊的所有參數，呼叫時只能依位置傳入，不能寫出引數名稱：   
## 2. * (僅限關鍵字引述keyword-only)
寫在單獨一個 * 右邊的所有參數，呼叫時必須寫出引數名稱，不能只靠位置傳入

```python
def calc_bmi(weight, height, /):
    bmi = weight / (height ** 2)
    print(f"BMI為{bmi:.1f}")

calc_bmi(65, 1.75)
```

```python
def send_email(to, *,subject, urgent=False):
    print(f"寄給{to},主旨:{subject},急件:{urgent}")

send_email("student@example.com", subject="課堂通知", urgent=True)
```

def 函式名稱 (僅限位置參數 , / , 位置或名稱即可, * , 僅限引數名稱):

```python
def register(id_no, /, name, *, role="student", gender, birthday):
    print(f"身分證字號:{id_no},姓名:{name},身分:{role},性別:{gender},生日:{birthday}")
register("A123456789", "小明", role="leader",gender="男", birthday="2000-01-01")
```

## *args 與 **kwargs — 定義函式時接收任意數量的引數 (打包 Packing)
有時候設計函式時，並不知道使用者會傳入幾個引數（例如計算任意個數字的總和，或接收額外的設定欄位）。
這時候可以在定義函式時，在參數前面加上 * 或 **，把多出來的引數「打包」起來。

### *args (一個星號) ：把多餘的「位置引數」打包成 tuple

### **kwargs (兩個星號) :把多餘的「引數名稱」打包成 dict

```python
def sum_scores(*scores):
    print("收到的資料 :", scores, type(scores))
    return sum(scores)

print(sum_scores(80, 90, 100))
print(sum_scores(80, 90))
print(sum_scores())
```

```python
def build_profile(name, **info):
    print("姓名:", name)
    print("其他資訊:", info, type(info))

build_profile("小華",city="台北", course="Python")
```
